# 🏥 Disease Prediction — Exploratory Data Analysis

This notebook provides interactive exploration of the three medical datasets:
- Heart Disease (UCI Cleveland)
- Diabetes (Pima Indians)
- Breast Cancer (Wisconsin Diagnostic)

It uses the `src.eda` module for generating visualizations.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import get_dataset, get_dataset_info
from src.preprocessing import handle_missing_values, encode_categoricals
from src.eda import run_eda

%matplotlib inline
sns.set_theme(style='whitegrid')
print('Setup complete!')
```

## 1. Dataset Overview

```python
datasets = ['heart_disease', 'diabetes', 'breast_cancer']

for ds in datasets:
    info = get_dataset_info(ds)
    print(f"\n{'='*50}")
    print(f"Dataset: {ds}")
    print(f"  Samples: {info['n_samples']}")
    print(f"  Features: {info['n_features']}")
    print(f"  Positive rate: {info['positive_rate']:.2%}")
    print(f"  Class distribution: {info['class_distribution']}")
```

## 2. Heart Disease Dataset

```python
X_heart, y_heart = get_dataset('heart_disease')
df_heart = pd.concat([X_heart, y_heart.rename('target')], axis=1)
print(f'Shape: {df_heart.shape}')
print(f'\nMissing values:\n{df_heart.isna().sum()}')
df_heart.describe()
```

```python
# Run full EDA suite for heart disease
heart_plots = run_eda('heart_disease')
for name, path in heart_plots.items():
    print(f'{name}: {path}')
```

## 3. Diabetes Dataset

```python
X_diab, y_diab = get_dataset('diabetes')
df_diab = pd.concat([X_diab, y_diab.rename('target')], axis=1)
print(f'Shape: {df_diab.shape}')
print(f'\nZero values in key columns:')
for col in ['glucose', 'blood_pressure', 'bmi', 'insulin', 'skin_thickness']:
    if col in df_diab.columns:
        print(f'  {col}: {(df_diab[col] == 0).sum()}')
df_diab.describe()
```

```python
diab_plots = run_eda('diabetes')
for name, path in diab_plots.items():
    print(f'{name}: {path}')
```

## 4. Breast Cancer Dataset

```python
X_bc, y_bc = get_dataset('breast_cancer')
df_bc = pd.concat([X_bc, y_bc.rename('target')], axis=1)
print(f'Shape: {df_bc.shape}')
print(f'\nFirst 10 features: {list(X_bc.columns[:10])}')
df_bc.describe()
```

```python
bc_plots = run_eda('breast_cancer')
for name, path in bc_plots.items():
    print(f'{name}: {path}')
```

## 5. Summary

Key observations:
- **Heart Disease**: Relatively balanced classes (~46% positive). 13 features with mix of continuous and categorical.
- **Diabetes**: Moderate imbalance (~35% positive). Notable zeros-as-missing pattern in glucose, blood pressure, BMI.
- **Breast Cancer**: 30 numeric features (mean, error, worst for 10 cell measurements). ~37% malignant.

All datasets benefit from feature scaling and class imbalance handling.
